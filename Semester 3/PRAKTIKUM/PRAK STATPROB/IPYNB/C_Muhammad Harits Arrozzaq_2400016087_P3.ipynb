```python
# Langkah-langkah Percobaan
def hitung_rata_rata (data):
    total = sum(data)
    rata_rata = total / len(data) 
    return rata_rata

data_tinggi_badan = [160, 165, 170, 175, 180]

rata_rata = hitung_rata_rata(data_tinggi_badan)

print("Rata-rata tinggi badan siswa adalah:", rata_rata, "cm")
```

```python
def hitung_median(data):
    data_urut = sorted(data)
    n = len(data_urut)
    if n % 2 == 0:
        median = (data_urut[n//2 - 1] + data_urut[n//2]) / 2
    else:
        median = data_urut[n//2]
    return median

data_nilai_ujian = [75, 80, 65, 90, 85, 70, 88, 92, 78]

median = hitung_median(data_nilai_ujian)
print("Median nilai ujian matematika siswa adalah:", median)
        
```

```python
#Langkah-langkah Praktikum
#Soal 1
import numpy as np

data = [
    5.2, 4.8, 6.0, 4.5, 5.7, 6.5, 4.9, 6.2, 5.8, 5.1,
    5.4, 5.3, 4.6, 6.3, 5.5, 4.7, 5.9, 6.1, 5.6, 4.4,
    6.4, 6.0, 5.0, 4.8, 6.2, 5.7, 5.1, 5.4, 5.8, 4.9,
    4.7, 5.5, 6.1, 6.3, 4.6, 5.2, 5.9, 5.6, 6.4, 6.0,
    5.3, 5.5, 6.1, 5.7, 5.8, 5.0, 5.9, 6.3, 5.1, 4.8
]


#1. Mean (rata-rata) penghasilan
mean = np.mean(data)

#2. Median (nilai tengah) penghasilan
median = np.median(data)

#3. Modus (nilai yang paling sering muncul) penghasilan
unik = list(set(data))
frekuensi = [data.count(i) for i in unik]
maks_frek = max(frekuensi)
modus = [unik[i] for i, f in enumerate(frekuensi) if f == maks_frek]

print(f"Mean penghasilan   : {mean:.2f}")
print(f"Median penghasilan : {median:.2f}")
print(f"Modus penghasilan  : {modus}")

#4. interpretasi
print("\nInterpretasi:")
print("Nilai mean dan median hampir sama → distribusi data simetris.")
print("Sebagian besar penghasilan berada di kisaran 5–6 juta rupiah.")
```

```python
#Soal 2
import numpy as np

data = [
    78, 85, 88, 90, 72, 75, 82, 87, 94, 96,
    65, 70, 80, 68, 85, 78, 92, 84, 91, 86,
    73, 79, 88, 94, 76, 89, 81, 83, 87, 90,
    63, 72, 85, 74, 77, 89, 86, 81, 83, 84
]

#1. Mean, Median, dan Modus
mean = np.mean(data)
median = np.median(data)
unik = list(set(data)) 
frekuensi = [data.count(i) for i in unik]
maks_frek = max(frekuensi)
modus = [unik[i] for i, f in enumerate(frekuensi) if f == maks_frek]

#2. Bandingkan hasil ketiga ukuran pemusatan data tersebut
if mean > median:
    bentuk = "miring ke kanan (positif skew)"
elif mean < median:
    bentuk = "miring ke kiri (negatif skew)"
else:
    bentuk = "simetris"

#3. Interpretasi
interpretasi = (
    "Sebagian besar nilai siswa berada di kisaran 80–90, "
    "menunjukkan performa siswa cukup baik secara keseluruhan."
)

print("_HASIL PERHITUNGAN NILAI UJIAN MATEMATIKA_")
print(f"Mean   : {mean:.2f}")
print(f"Median : {median:.2f}")
print(f"Modus  : {modus}")
print(f"\nDistribusi data: {bentuk}")
print("\nInterpretasi:")
print(interpretasi)

```

```python
#Soal 3
import numpy as np

data = [
    450, 420, 475, 500, 490, 460, 430, 510, 470, 480, 425, 440, 520, 450, 485, 510, 500, 460, 445,
    455, 480, 525, 495, 470, 520, 500, 435, 480, 505, 490, 460, 475, 525, 455, 515, 500, 490, 510,
    495, 525, 530, 460, 470, 480, 440, 450, 505, 510, 480, 495, 500, 490, 525, 460, 510, 470, 455,
    500, 490, 475
]

#1. Mean harga rumah
mean = np.mean(data)

#2. Median harga rumah
median = np.median(data)

#3. Modus harga rumah
unik = list(set(data))
frekuensi = [data.count(i) for i in unik]
maks_frek = max(frekuensi)
modus = [unik[i] for i, f in enumerate(frekuensi) if f == maks_frek]

print(f"Mean   : {mean:.2f}")
print(f"Median : {median:.2f}")
print(f"Modus  : {modus}")

if mean > median:
    print("Distribusi: miring ke kanan (banyak harga tinggi).")
elif mean < median:
    print("Distribusi: miring ke kiri (banyak harga rendah).")
else:
    print("Distribusi: simetris (data merata).")

print("\nAnalisis:")
print(f"Harga rumah rata-rata sekitar {mean:.0f} ratus juta rupiah.")
print("Sebagian besar harga rumah berada di sekitar nilai median, menunjukkan konsentrasi di rentang menengah.")

```
