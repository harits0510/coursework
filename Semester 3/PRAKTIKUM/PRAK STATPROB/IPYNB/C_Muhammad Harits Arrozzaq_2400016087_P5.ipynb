```python
#Langkah-langkah Percobaan
import numpy as np

# Data nilai ujian matematika siswa
data_nilai_ujian = np.array([65, 70, 72, 75, 78, 80, 82, 85, 88, 92])

# Hitung kuartil menggunakan NumPy
kuartil = np.percentile(data_nilai_ujian, [25, 50, 75])
q1, q2, q3 = kuartil[0], kuartil[1], kuartil[2]

print("Kuartil pertama (Q1):", q1)
print("Kuartil kedua (Q2 atau median): ", q2)
print("Kuartil ketiga (Q3):", q3)
```

```python
import statistics

# Data nilai ujian matematika siswa
data_nilai_ujian = [65, 70, 72, 75, 78, 80, 82, 85, 88, 92]

# Hitung kuartil menggunakan statistics
q1 = statistics.quantiles(data_nilai_ujian, n=4) [0]
q2 = statistics.median(data_nilai_ujian)
q3 = statistics.quantiles(data_nilai_ujian, n=4) [-1]

print("Kuartil pertama (Q1):", q1)
print ("Kuartil kedua (Q2 atau median): ", q2)
print ("Kuartil ketiga (Q3):", q3)
```

```python
import pandas as pd

# Data nilai ujian matematika siswa
data_nilai_ujian = pd.Series([65, 70, 72, 75, 78, 80, 82, 85, 88, 92])

# Hitung kuartil menggunakan Pandas
q1 = data_nilai_ujian.quantile (0.25)
q2 = data_nilai_ujian.median ()
q3 = data_nilai_ujian.quantile (0.75)

print ("Kuartil pertama (Q1):", q1)
print("Kuartil kedua (Q2 atau median):", q2)
print ("Kuartil ketiga (Q3):", q3)
```

```python
# Langkah-langkah Praktikum
import numpy as np

# Dataset nilai ujian
data = np.array([
    56, 70, 68, 72, 80, 85, 90, 60, 62, 75, 68, 70, 74, 77, 83, 88,
    92, 55, 65, 78, 67, 72, 76, 81, 87, 93, 66, 73, 79, 85, 69, 71,
    84, 90, 95, 57, 64, 72, 82, 91
])

# Urutkan data
data = np.sort(data)

# Kuartil
Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

# Desil
D5 = np.percentile(data, 50)
D8 = np.percentile(data, 80)

# Persentil
P20 = np.percentile(data, 20)
P50 = np.percentile(data, 50)
P90 = np.percentile(data, 90)

# Hasil
print("Kuartil 1 (Q1) =", Q1)
print("Kuartil 2 (Median) =", Q2)
print("Kuartil 3 (Q3) =", Q3)
print("Desil ke-5 (D5) =", D5)
print("Desil ke-8 (D8) =", D8)
print("Persentil ke-20 (P50) =", P20)
print("Persentil ke-50 (P50) =", P50)
print("Persentil ke-90 (P90) =", P90)
```
