```python
import matplotlib.pyplot as plt 
import numpy as np

data_korelasi = np.array([[0.8, 0.2, 0.5, 0.9],
                            [0.6, 0.1, 0.3, 0.7],
                            [0.3, 0.6, 0.4, 0.81]])

plt.imshow(data_korelasi, cmap='viridis', interpolation='nearest')

plt.xticks (np.arange(len(data_korelasi[0])), ['A1', 'A2', 'A3', 'A4'])
plt.yticks(np.arange(len(data_korelasi)), ['B1', 'B2', 'B3'])

for i in range(len(data_korelasi)): 
    for j in range(len (data_korelasi[0])):
        plt.text(j, i, '{:.2f}'. format(data_korelasi[i, j]),
                 ha='center', va='center', color='white')

plt.title('Heatmap Korelasi antara Variabel A dan B') 
plt.xlabel('Variabel A')
plt.ylabel('Variabel B')

plt.colorbar()
plt.show()
```

```python
import matplotlib.pyplot as plt

keluhan = {
    'Keterlambatan Pengiriman': 35,
    'Produk Tidak Sesuai Pesanan': 20,
    'Kualitas Produk Rendah': 15,
    'Pelayanan Kurang Memuaskan': 10,
    'Lainnya': 5
}

keluhan_sorted = dict(sorted(keluhan.items(), key=lambda item: item[1], reverse=True))

total_frekuensi = sum(keluhan. values ())

persentase_kumulatif = 0
persentase_kumulatif_list = []
for value in keluhan_sorted. values():
    persentase = (value / total_frekuensi) * 100
    persentase_kumulatif += persentase
    persentase_kumulatif_list.append (persentase_kumulatif)

fig, ax1 = plt.subplots()

ax1.bar(keluhan_sorted.keys(), keluhan_sorted. values(), color='b')
ax1.set_xlabel('Jenis Keluhan')
ax1.set_ylabel('Frekuensi', color='b')

ax2 = ax1.twinx()
ax2.plot(keluhan_sorted.keys(), persentase_kumulatif_list, color='r', marker='o')
ax2.set_ylabel('Persentase Kumulatif', color='r')

plt.title('Diagram Pareto Keluhan Pelanggan')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Data tinggi badan (cm)
data = [
    165, 170, 172, 160, 175, 168, 180, 174, 169, 177,
    162, 167, 173, 178, 164, 171, 176, 166, 182, 159,
    161, 172, 169, 163, 179, 177, 158, 180, 170, 173,
    174, 168, 171, 178, 166, 160, 181, 165, 167, 176,
    159, 183, 161, 175, 169, 172, 164, 162, 174, 173,
    171, 178, 164, 167, 181, 166, 170, 177, 180, 176,
    182, 175, 161, 160, 165, 168, 172, 170, 179, 177,
    162, 169, 174, 168, 173, 176, 182, 165, 159, 171
]

# Konversi ke DataFrame
df = pd.DataFrame({'Tinggi': data})

# Jumlah kelas (rumus Sturges)
k = round(1 + 3.3 * np.log10(len(df)))

# Rentang dan lebar kelas
x_min, x_max = df['Tinggi'].min(), df['Tinggi'].max()
interval = int(np.ceil((x_max - x_min) / k))

# Interval kelas & distribusi frekuensi
bins = np.arange(x_min, x_max + interval, interval)
freq, edges = np.histogram(df['Tinggi'], bins=bins)

# Tabel distribusi
tabel = pd.DataFrame({
    'Interval Kelas': [f'{int(edges[i])}-{int(edges[i+1]-1)}' for i in range(len(freq))],
    'Frekuensi': freq
})

print("___Tabel Distribusi Frekuensi Tinggi Badan___")
print(tabel)

# Histogram
plt.hist(df['Tinggi'], bins=bins, color='skyblue', edgecolor='black')
plt.title('Histogram Tinggi Badan (cm)')
plt.xlabel('Tinggi Badan (cm)')
plt.ylabel('Frekuensi')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


# Interpretasi
print(" ___Interpretasi___ ")
print("distribusi data tinggi badan terlihat mendekati simetris dengan puncak di sekitar 170–177 cm.")
print("sebagian besar data berada di kisaran tersebut, dan tidak terdapat outlier ekstrem.")
print("distribusi ini normal dengan sedikit kemiringan ke kanan.")
```

```python
import matplotlib.pyplot as plt

# Data
jenis_minuman = ['Teh', 'Kopi', 'Jus', 'Air Putih', 'Soda']
frekuensi = [25, 30, 15, 20, 10]

# Warna baru: merah, kuning, hijau, skyblue, (tambahan warna opsional untuk ke-5)
warna = ['red', 'yellow', 'green', 'skyblue', 'orange']

# Diagram Batang
plt.figure(figsize=(8, 6))
plt.bar(jenis_minuman, frekuensi, color=warna)
plt.title('Diagram Batang Preferensi Minuman (100 Responden)')
plt.xlabel('Jenis Minuman')
plt.ylabel('Frekuensi (Jumlah Responden)')
plt.show()

# Diagram Lingkaran
plt.figure(figsize=(6, 6))
plt.pie(frekuensi, labels=jenis_minuman, colors=warna, autopct='%1.1f%%', startangle=90)
plt.title('Diagram Lingkaran Preferensi Minuman')
plt.show()

# Interpretasi
print("___Interpretasi___")
print("Berdasarkan hasil visualisasi:")
print("Minuman yang paling populer adalah Kopi dengan 30 responden (30%).")
print("Minuman yang paling tidak populer adalah Soda dengan 10 responden (10%).")
print("Minuman Teh cukup diminati dengan 25 responden (25%), diikuti oleh Air Putih 20 responden (20%) dan Jus 15 responden (15%).")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Dataset Berat Badan(Kg) dan Tinggi Badan(Cm) dari 50 orang
data = {
    "Berat Badan (kg)": [
        55, 60, 65, 70, 58, 62, 75, 80, 68, 59,
        64, 70, 73, 66, 78, 85, 90, 72, 76, 81,
        50, 53, 56, 63, 66, 74, 82, 67, 71, 77,
        58, 64, 68, 75, 69, 73, 81, 78, 88, 83,
        79, 65, 63, 55, 60, 57, 70, 80, 86, 90
    ],
    "Tinggi Badan (cm)": [
        160, 162, 165, 170, 163, 167, 180, 175, 172, 161,
        168, 173, 176, 171, 177, 183, 185, 174, 179, 181,
        155, 158, 161, 169, 170, 178, 180, 172, 175, 182,
        164, 169, 173, 179, 170, 177, 182, 178, 184, 180,
        177, 167, 166, 159, 162, 165, 172, 181, 185, 188
    ]
}

# Konversi ke DataFrame
df = pd.DataFrame(data)

# Scatter Plot
plt.figure(figsize=(8, 6))
plt.scatter(df["Tinggi Badan (cm)"], df["Berat Badan (kg)"],
            color='skyblue', edgecolor='black')
plt.title("Scatter Plot Hubungan Tinggi Badan dan Berat Badan (50 Orang)")
plt.xlabel("Tinggi Badan (cm)")
plt.ylabel("Berat Badan (kg)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Hitung Korelasi
korelasi = df["Tinggi Badan (cm)"].corr(df["Berat Badan (kg)"])

# Interpretasi
print("___Interpretasi___")
print("1. Pola Hubungan:")
print("Titik-titik pada scatter plot membentuk pola naik — semakin tinggi seseorang, semakin besar berat badannya.")
print()
print("2. Jenis Korelasi:")
print("Terdapat korelasi positif kuat antara tinggi dan berat badan (r = 0,97).")
print()
print("3. Kesimpulan:")
print("Semakin tinggi seseorang, umumnya berat badannya juga meningkat.")
```
