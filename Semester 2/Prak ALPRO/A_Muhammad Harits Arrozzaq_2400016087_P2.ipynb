```python
def binsearch(data, key):
    awal = 0
    akhir = len(data) - 1
    ketemu = False

    while (awal <= akhir) and not ketemu:
        tengah = (awal + akhir) // 2

        if key == data[tengah]:
            ketemu = True
            return f"Data {key} ditemukan di posisi {tengah}"
        elif key < data[tengah]:
            akhir = tengah - 1
        else:
            awal = tengah + 1

    return "Data tidak ditemukan"

data = sorted([57, 88, 77, 26, 96])
key = 88
hasil = binsearch(data, key)
print(hasil)
```

```python
def sqesearch(data, key):
    pos = []
    iterations = 0
    for i in range(len(data)):
        iterations += 1
        if data[i] == key:
            pos.append(i + 1)
    if len(pos) > 0:
        print("Data", key, "sebanyak", len(pos), "ditemukan di posisi", pos)
    else:
        print("Data tidak ditemukan")
    print("Jumlah iterasi (Sequential Search):", iterations)
    return pos

def binsearch(data, key):
    data = sorted(data)
    awal = 0
    akhir = len(data) - 1
    ketemu = False
    iterations = 0

    while (awal <= akhir) and not ketemu:
        iterations += 1
        tengah = (awal + akhir) // 2

        if key == data[tengah]:
            ketemu = True
            print(f"Data {key} ditemukan di posisi {tengah+1}")

        elif key < data[tengah]:
            akhir = tengah - 1
        else:
            awal = tengah + 1


    print("Jumlah iterasi (Binary Search):", iterations)
    if not ketemu:
        return "Data tidak ditemukan"


text = "sistem informasi"
key = "i"


print("Sequential Search:")
A = sqesearch(text, key)
print(A)


print("\nBinary Search:")
hasil = binsearch(text, key)
```
