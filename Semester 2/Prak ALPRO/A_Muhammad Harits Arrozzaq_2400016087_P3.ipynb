```python
#modifikasi ungsi untuk mengur
def descending_insertion_sort(sort_list):
    for i in range(1, len(sort_list)):
        key = sort_list[i]
        j = i - 1
        while j >= 0 and key > sort_list[j]:
            sort_list[j + 1] = sort_list[j]
            j -= 1
        sort_list[j + 1] = key
        print('iterasi', i - 1, sort_list)

A = [4, 3, 5, 6, 2, 78, 98]
descending_insertion_sort(A.copy())
```

```python
#Modifikasi fungsi sehingga menampilkan hasil tiap langkah pengurutan
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

        print(f"iterasi {i - 1} {arr}")

A = [4, 3, 5, 6, 2, 78, 98]
insertion_sort(A)
```
