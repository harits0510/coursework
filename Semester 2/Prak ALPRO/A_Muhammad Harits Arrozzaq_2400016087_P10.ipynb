```python
class alat_elektronik:
    jumlah=0
    def __init__(unit, nama, merk, status):
        unit.nama=nama
        unit.merk=merk
        unit.status=status
        alat_elektronik.jumlah +=1

    def tampil_jumlah(unit):
        print("jumlah_unit : ", alat_elektronik.jumlah)

    def tampil_alat_elektronik(unit):
        print("nama : ", unit.nama, ", merk : ", unit.merk, ", status : ", unit.status)

unit1=alat_elektronik("Televisi", "Samsung", "Hidup")
unit2=alat_elektronik("AC", "LG", "Mati")
unit3=alat_elektronik("Handphone", "POCO", "Hidup")
unit4=alat_elektronik("Laptop", "LOQ", "Mati")

unit1.tampil_alat_elektronik()
unit2.tampil_alat_elektronik()
unit3.tampil_alat_elektronik()
unit4.tampil_alat_elektronik()

alat_elektronik.jumlah
```
