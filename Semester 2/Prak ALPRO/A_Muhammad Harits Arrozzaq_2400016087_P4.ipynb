```python
def Selection_Sort(A):
    for i in range(len(A) - 1):
        min_idx = i
        for j in range(i + 1, len(A)):
            if A[min_idx] > A[j]:
                min_idx = j
        temp = A[i]
        A[i] = A[min_idx]
        A[min_idx] = temp
        print('Iterasi', i + 1, ':', A)

A = [99, 88, 55, 22, 11, 77, 66, 33, 44]
Selection_Sort(A)
```

```python
def Selection_Sort_Descending(A):
    n = len(A)
    for i in range(n - 1):
        max_index = i
        for j in range(i + 1, n):
            if A[j] > A[max_index]:
                max_index = j
                
        A[i], A[max_index] = A[max_index], A[i]
        
        print(f"indek key : {i} tukar : {max_index}")
        print(f"iterasi {i} {A}")

# Contoh penggunaan
A = [99, 88, 55, 22, 11, 77, 66, 33, 44]
print(f"A={A}")
Selection_Sort_Descending(A)
```
