```python
#nomor 1
#fungsi rekursif
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
#nomor 1
#fungsi fibonacci dengan iterasi (for)
def fib_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fib_iter(5))
```

```python
#nomor 1
#fungsi fibonacci dengan iterasi (while)
def fib_while(n):
    a, b = 0, 1
    count = 0
    while count < n:
        a, b = b, a + b
        count += 1
    return a

print(fib_while(5))  
```

```python
#nomor 2A
a.analisis kasus penyetop dan rekursinya
Kasus penyetop (base case):

Jika x == 0, maka 2^0 = 1.

Rekursi (recurrence relation):

2^x = 2 * 2^(x-1)
```

```python
#nomor 2B
#program menggunakan rekursif
def power_of_two_recursive(x):
    if x == 0:
        return 1
    else:
        return 2 * power_of_two_recursive(x - 1)

print(power_of_two_recursive(5))
```

```python
#nomor 2C
#program dengan iterasi/perulangan (for)
def power_of_two_for(x):
    result = 1
    for i in range(x):
        result *= 2
        print(f"Step {i+1}: result = {result}")
    return result

print(power_of_two_for(5))
```

```python
#nomor 2C
#program dengan iterasi/perulangan (while)
def power_of_two_while(x):
    result = 1
    count = 0
    while count < x:
        result *= 2
        count += 1
        print(f"Step {count}: result = {result}")
    return result

print(power_of_two_while(5))
```
