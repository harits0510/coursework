```python
def bubble2(A):
    n = len(A)
    for i in range(n):
        for j in range(0, n - i - 1):
            if A[j] < A[j + 1]:  # Ubah menjadi > jika pengurutan data secara ascending
                print(f"tukar {A[j]} dengan {A[j+1]}")
                A[j], A[j + 1] = A[j + 1], A[j]
                print(f"urutan menjadi {A}")
    return A

# Pemanggilan fungsi
A = [3, 1, 8, 4, 2]
bubble2(A)
```
