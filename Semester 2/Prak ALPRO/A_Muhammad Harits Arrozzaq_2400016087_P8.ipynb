```python
# Presensi

# daftar nama mahasiswa
mahasiswa = [
    "Yunus", "Aji", "Farhan", "Dama", "Frizi",
    "Al", "Harits"
]

# Inisialisasi penghitung kehadiran
hadir = 0
tidak_hadir = 0

# Proses input kehadiran untuk setiap mahasiswa
for nama in mahasiswa:
    while True:
        status = input(f"Apakah {nama} hadir? (y/n): ").strip().lower()
        if status == 'y':
            hadir += 1
            break
        elif status == 'n':
            tidak_hadir += 1
            break
        else:
            print("Input harus 'y' (hadir) atau 'n' (tidak hadir). Silakan ulangi.")

# Menampilkan rekap hasil kehadiran
print("\n--- Rekap Kehadiran ---")
print("Total siswa      :", len(mahasiswa))
print("Jumlah yang hadir:", hadir)
print("Jumlah tidak hadir:", tidak_hadir)
```

```python
# Kalkulator Pengembalian Investasi

# Meminta input dari pengguna
total_biaya = float(input("Masukkan total biaya/investasi (dalam satuan mata uang): "))
total_penjualan = float(input("Masukkan total penjualan (dalam satuan mata uang): "))

# Menghitung laba
laba = total_penjualan - total_biaya

# Menampilkan hasil laba/rugi
if laba > 0:
    print("Bisnis untung.")
elif laba < 0:
    print("Bisnis rugi.")
else:
    print("Bisnis impas.")

# Menghitung ROI
if total_biaya != 0:
    roi = (laba / total_biaya) * 100
    print(f"Return On Investment (ROI) bisnis Anda adalah: {roi:.2f}%")
else:
    print("Tidak dapat menghitung ROI karena total biaya adalah 0.")
```
