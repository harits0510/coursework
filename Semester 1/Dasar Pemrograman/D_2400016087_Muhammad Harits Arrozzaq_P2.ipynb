```python
#Tipe Data String
##Menyatakan rangkaian karakter, diapit tanda kutip tunggal atau ganda
text="Belajar Python sangat mudah"
print(text)
```

```python
#Tipe Data Integer
##Menyatakan bilangan bulat
angka=1234
print(angka)
```

```python
#Tipe Data Float
##Menyatakan bilangan dengan titik desimal
pecahan=50.0609
print(pecahan)
```

```python
#Tipe Data Complex Number
##Menyatakan bilangan kompleks dengan bagian real dan imajiner
komplek=10j
print(komplek)
```

```python
#Tipe Data Boolean
##Menyatakan nilai kebenaran,hanya ada dua kemungkinan: True atau False
bin=True 
print(bin)
```
