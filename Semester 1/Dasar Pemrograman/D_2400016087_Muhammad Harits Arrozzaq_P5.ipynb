```python
N = int(input("Masukkan N: "))
total = 0
angka_list = []

for i in range(N):
    angka = int(input(f"Masukkan angka ke-{i+1}: "))
    angka_list.append(angka)
    total += angka

hasil = " + ".join(map(str, angka_list))
print(f"{hasil} = {total}")
```

```python
N = int(input("Masukkan N: "))
total = 0

for i in range(N):
    angka = int(input(f"Masukkan angka ke-{i+1}: "))
    total += angka

rata_rata = total / N

if rata_rata > 50:
    print("Rata-rata di atas 50")
else:
    print("Rata-rata di bawah atau sama dengan 50")
```

```python
N = int(input("Masukkan N: "))
genap_count = 0
ganjil_count = 0

for i in range(N):
    angka = int(input(f"Masukkan angka ke-{i+1}: "))
    if angka % 2 == 0:
        genap_count += 1
    else:
        ganjil_count += 1

if genap_count == 0:
    print("Tidak ada bilangan genap.")
else:
    print(f"Jumlah bilangan genap: {genap_count}")

if ganjil_count == 0:
    print("Tidak ada bilangan ganjil.")
else:
    print(f"Jumlah bilangan ganjil: {ganjil_count}")
```
