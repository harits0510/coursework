```python
def max_bilangan(x, y, z):
    if x >= y and x >= z:
        return x
    elif y >= x and y >= z:
        return y
    else:
        return z

x = int(input("masukkan nilai x: "))
y = int(input("masukkan nilai y: "))
z = int(input("masukkan nilai z: "))

maksimum = max_bilangan(x, y, z)
print("nilai maksimum adalah:",maksimum)
```

```python
def min_bilangan(x, y, z):
    if x <= y and x <= z:
        return x
    elif y <= x and y <= z:
        return y
    else:
        return z

x = int(input("masukkan nilai x: "))
y = int(input("masukkan nilai y: "))
z = int(input("masukkan nilai z: "))

minimum = min_bilangan(x, y, z)
print("nilai minimum adalah:",minimum)
```

```python
def hitung_bmi(berat, tinggi_cm):
  """Menghitung BMI dan mengklasifikasikannya ke dalam kategori.
      
    berat: Berat badan dalam kilogram.
    tinggi_cm: Tinggi badan dalam sentimeter.

  Returns:
    String yang berisi nilai BMI dan kategori BMI.
  """
  tinggi_m = tinggi_cm / 100
  bmi = berat / (tinggi_m ** 2)

  if bmi < 18.5:
    kategori = "Kurang berat badan"
  elif 18.5 <= bmi < 24.9:
    kategori = "Normal"
  elif 25 <= bmi < 29.9:
    kategori = "Kelebihan berat badan"
  else:
    kategori = "Obesitas"
  return f"BMI Anda adalah {bmi:.2f}, termasuk kategori {kategori}"
berat = float(input("Masukkan berat badan Anda (kg): "))
tinggi_cm = float(input("Masukkan tinggi badan Anda (cm): "))


print(hitung_bmi(berat, tinggi_cm))
```
