```python
kereta_api = [
    {"ID": 1, "Nama": "Express A", "Asal": "City X", "Tujuan": "City Y", "Kursi": 50, "Harga": 150},
    {"ID": 2, "Nama": "Express B", "Asal": "City X", "Tujuan": "City Z", "Kursi": 30, "Harga": 200},
    {"ID": 3, "Nama": "Express C", "Asal": "City X", "Tujuan": "City Z", "Kursi": 0, "Harga": 250},
]

tiket_dipesan = []

def tampilkan_daftar_kereta():
    print("\nDaftar Kereta:")
    for kereta in kereta_api:
        print(f"ID: {kereta['ID']}, Nama: {kereta['Nama']}, From: {kereta['Asal']}, To: {kereta['Tujuan']}, Seats: {kereta['Kursi']}, Price: {kereta['Harga']}")

def pesan_tiket():
    try:
        id_kereta = int(input("\nMasukkan ID Kereta: "))
        kereta = next((k for k in kereta_api if k["ID"] == id_kereta), None)

        if kereta:
            if kereta["Kursi"] > 0:
                nama_penumpang = input("Masukkan Nama Penumpang: ")
                kereta["Kursi"] -= 1
                tiket = {
                    "ID Kereta": kereta["ID"],
                    "Nama Penumpang": nama_penumpang,
                    "Nama Kereta": kereta["Nama"],
                    "Harga": kereta["Harga"]
                }
                tiket_dipesan.append(tiket)
                print(f"\nTiket berhasil dipesan untuk {nama_penumpang} di kereta {kereta['Nama']}.")
            else:
                print("\nMaaf, kursi pada kereta ini sudah penuh.")
        else:
            print("\nID Kereta tidak ditemukan.")
    except ValueError:
        print("\nInput tidak valid. Harap masukkan angka.")

def lihat_tiket():
    print("\nTiket yang sudah dipesan:")
    if tiket_dipesan:
        for tiket in tiket_dipesan:
            print(f"Nama Penumpang: {tiket['Nama Penumpang']}, Nama Kereta: {tiket['Nama Kereta']}, Harga: {tiket['Harga']}")
    else:
        print("Belum ada tiket yang dipesan.")

def keluar():
    print("\nTerima kasih telah menggunakan sistem pemesanan tiket. Sampai jumpa!")
    exit()

def menu():
    while True:
        print("\nMenu:")
        print("1. Menampilkan Daftar Kereta")
        print("2. Memesan Tiket")
        print("3. Melihat Tiket yang Sudah Dipesan")
        print("4. Keluar")
        try:
            pilihan = int(input("Pilih menu: "))
            if pilihan == 1:
                tampilkan_daftar_kereta()
            elif pilihan == 2:
                pesan_tiket()
            elif pilihan == 3:
                lihat_tiket()
            elif pilihan == 4:
                keluar()
            else:
                print("\nPilihan menu tidak valid.")
        except ValueError:
            print("\nInput tidak valid. Harap masukkan angka.")

menu()
```

```python
import random

def print_welcome():
    print("Selamat datang di Game Tebak Angka!")
    print("Aturan permainan:")
    print("1. Pemain harus menebak angka yang diacak dalam rentang tertentu.")
    print("2. Pemain memiliki maksimal 5 kesempatan untuk menebak.")
    print("3. Setelah setiap tebakan, Pemain akan diberi tahu apakah tebakan terlalu tinggi atau terlalu rendah.")
    print("4. Jika Pemain menebak dengan benar, game selesai.")

def choose_level():
    print("\nPilih tingkat kesulitan:")
    print("1. Level 1: Tebak angka antara 1-10")
    print("2. Level 2: Tebak angka antara 1-15")
    print("3. Level 3: Tebak angka antara 1-20")
    print("4. Level 4: Tentukan rentang angka kustom")

    while True:
        try:
            choice = int(input("Masukkan pilihan level Anda (1-4): "))
            if choice == 1:
                return 1, 10
            elif choice == 2:
                return 1, 15
            elif choice == 3:
                return 1, 20
            elif choice == 4:
                while True:
                    try:
                        min_range = int(input("Masukkan angka minimum: "))
                        max_range = int(input("Masukkan angka maksimum: "))
                        if min_range < max_range:
                            return min_range, max_range
                        else:
                            print("Rentang tidak valid. Pastikan angka minimum lebih kecil dari angka maksimum.")
                    except ValueError:
                        print("Harap masukkan angka yang valid.")
            else:
                print("Pilihan tidak valid. Silakan pilih 1-4.")
        except ValueError:
            print("Masukkan angka valid antara 1-4.")

def play_game():
    print_welcome()
    min_range, max_range = choose_level()
    number_to_guess = random.randint(min_range, max_range)
    max_attempts = 5

    print(f"\nSaya telah memilih angka antara {min_range} dan {max_range}. Anda memiliki {max_attempts} kesempatan untuk menebak.")

    for attempt in range(1, max_attempts + 1):
        while True:
            try:
                guess = int(input(f"\nTebakan ke-{attempt}: Masukkan angka Anda: "))
                break
            except ValueError:
                print("Harap masukkan angka yang valid.")

        if guess == number_to_guess:
            print(f"Selamat! Anda menebak dengan benar dalam {attempt} percobaan.")
            return
        elif guess < number_to_guess:
            print("Tebakan Anda terlalu rendah.")
        else:
            print("Tebakan Anda terlalu tinggi.")

    print(f"\nSayang sekali! Anda telah kehabisan kesempatan. Angka yang saya pilih adalah {number_to_guess}.")

def main():
    while True:
        play_game()
        play_again = input("\nApakah Anda ingin bermain lagi? (y/n): ").lower()
        if play_again != 'y':
            print("Terima kasih telah bermain! Sampai jumpa!")
            break

if __name__ == "__main__":
    main()
```
