```python
for x in range(1, 6):
    for y in range(1, 6):  
        if x * y <= 12:  
            print(f"({x}, {y})", end=" ")
    print() 
```

```python
x = int(input("Masukkan angka x: "))
y = int(input("Masukkan angka y: "))

if x % y == 0:
    print(f"{x} adalah kelipatan dari {y}")
else:
    print(f"{x} bukan kelipatan dari {y}")

print(f"Kelipatan dari {y} hingga {x}:")
for i in range(y, x + 1, y):
    print(i, end=" ")
```
