```python
##Program untuk menentukan jenis kendraan

#Program menggunakan if-elif-else
kode_kendaraan = input("Masukkan kode kendaraan (M untuk motor, MOB untuk mobil, B untuk bus): ")

if kode_kendaraan == 'M':
    jenis_kendaraan = 'Motor'
elif kode_kendaraan == 'MOB':
    jenis_kendaraan = 'Mobil'
elif kode_kendaraan == 'B':
    jenis_kendaraan = 'Bus'
else:
    jenis_kendaraan = 'Kode kendaraan tidak valid'

print("Jenis kendaraan:", jenis_kendaraan)
```

```python
#Program menggunakan dictionary
kode_kendaraan = input("Masukkan kode kendaraan (M untuk motor, MOB untuk mobil, B untuk bus): ")

kendaraan_dict = {
    'M': 'Motor',
    'MOB': 'Mobil',
    'B': 'Bus'
}

jenis_kendaraan = kendaraan_dict.get(kode_kendaraan, 'Kode kendaraan tidak valid')

print("Jenis kendaraan:", jenis_kendaraan)
```

```python
##Program untuk menampilkan nama musim
#Program menggunakan if-elif-else
nomor_musim = int(input("Masukkan nomor musim (1 untuk Musim Semi, 2 untuk Musim Panas, 3 untuk Musim Gugur, 4 untuk Musim Dingin): "))

if nomor_musim == 1:
    nama_musim = 'Musim Semi'
elif nomor_musim == 2:
    nama_musim = 'Musim Panas'
elif nomor_musim == 3:
    nama_musim = 'Musim Gugur'
elif nomor_musim == 4:
    nama_musim = 'Musim Dingin'
else:
    nama_musim = 'Nomor musim tidak valid'

print("Nama musim:", nama_musim)
```

```python
#Program menggunakan dictionary
nomor_musim = int(input("Masukkan nomor musim (1 untuk Musim Semi, 2 untuk Musim Panas, 3 untuk Musim Gugur, 4 untuk Musim Dingin): "))

musim_dict = {
    1: 'Musim Semi',
    2: 'Musim Panas',
    3: 'Musim Gugur',
    4: 'Musim Dingin'
}

nama_musim = musim_dict.get(nomor_musim, 'Nomor musim tidak valid')

print("Nama musim:", nama_musim)
```

```python
##Program menentukan kategori tiket berdasarkan kelas
# Input kode kelas tiket
kode_kelas = input("Masukkan kode kelas tiket (EK untuk Ekonomi, BS untuk Bisnis, FR untuk First Class): ")

kategori_tiket = {
    'EK': 'Ekonomi',
    'BS': 'Bisnis',
    'FR': 'First Class'
}

kategori = kategori_tiket.get(kode_kelas, 'Kode kelas tidak valid')

print("Kategori tiket:", kategori)
```
