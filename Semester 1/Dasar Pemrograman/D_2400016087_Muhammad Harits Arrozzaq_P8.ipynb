```python
#Mencari elemen dalam array

def cari_elemen(array, elemen):
    if elemen in array:
        index = array.index(elemen)
        return f"Elemen {elemen} ditemukan pada indeks {index}."
    else:
        return f"Elemen {elemen} tidak ditemukan dalam array."

array = [10, 20, 30, 40, 50]
elemen = 50
print(cari_elemen(array, elemen))
```

```python
#Revers Array

def reverse_array(array):
    return array[::-1]

array = [1, 2, 3, 4, 5]
print("Array asli:", array)
print("Array terbalik:", reverse_array(array))
```

```python
#Sortir Array

def bubble_sort(array, ascending=True):
    n = len(array)
    for i in range(n):
        for j in range(0, n-i-1):
            if (ascending and array[j] > array[j+1]) or (not ascending and array[j] < array[j+1]):
                array[j], array[j+1] = array[j+1], array[j]
    return array

array = [5, 3, 8, 6, 2]
print("Array sebelum disortir:", array)
print("Ascending:", bubble_sort(array.copy(), ascending=True))
print("Descending:", bubble_sort(array.copy(), ascending=False))
```

```python
#Program pencarian duplikat dalam array

def cari_duplikat(array):
    duplikat = []
    unik = set()
    for elemen in array:
        if elemen in unik and elemen not in duplikat:
            duplikat.append(elemen)
        else:
            unik.add(elemen)
    return duplikat

array = [1, 2, 3, 4, 5, 2, 3]
print("Array:", array)
print("Elemen duplikat:", cari_duplikat(array))
```

```python
#Array dengan string

def string_panjang_lebih_dari(array, panjang_minimal):
    hasil = [string for string in array if len(string) > panjang_minimal]
    return hasil

array = ["Alice", "Bob", "Charlie", "Diana"]
panjang_minimal = 4
print(f"Nama dengan panjang lebih dari {panjang_minimal} karakter:", string_panjang_lebih_dari(array, panjang_minimal))
```
